```python
## 클래스 속성 사용하기
# class 클래스이름:
#    속성=값

class Person:
    bag = []

    def put_bag(self, stuff):
        self.bag.append(stuff)

james = Person()
james.put_bag('책')

maria = Person()
maria.put_bag('열쇠')

print(james.bag)
print(maria.bag)
```

```python
## 인스턴스 속성 사용하기
class Person:
    def __init__(self):
        self.bag = []

    def put_bag(self, stuff):
        self.bag.append(stuff)

james = Person()
james.put_bag('책')

maria = Person()
maria.put_bag('열쇠')

print(james.bag)
print(maria.bag)
```

```python
## 비공개 클래스 속성 사용하기
# class 클래스이름:
#     __속성 = 값

class Knight:
    __item_limit = 10

    def print_item_limit(self):
        print(Knight.__item_limit)

x = Knight()
x.print_item_limit()

x.__item_limit     # 비공개 클래스이기 때문에 안에서는 작동이 되는데 밖에서는 노출이 안됨
```

```python
## 정적 메서드
# class 클래스이름:
#     @staticmethod
#     def 메서드(매개변수1, 매개변수2):
#         코드
class Calc:
    @staticmethod
    def add(a, b):
        print(a + b)

    @staticmethod
    def mul(a, b):
        print(a * b)

Calc.add(10, 20)
Calc.mul(10, 20)


```

```python
## 클래스 매서드 사용하기
# class 클래스이름:
#     @classmethod
#     def 메서드(cls, 매개변수1, 매개변수2):
#         코드

class Person:
    count = 0

    def __init__(self):
        Person.count += 3

    @classmethod
    def print_count(cls):
        print(f'{cls.count}명 생성됭었습니다')

james = Person()
maria = Person()

Person.print_count()
```

```python
#35.5
class Date:
    @staticmethod
    def is_date_valid(date_string):
        year, month, day = map(int, date_string.split('-'))
        return month <= 12 and day <= 31

if Date.is_date_valid('2000-10-31'):
    print('올바른 날짜입니다.')
else:
    print('잘못된 날짜 형식입니다.')
```

```python
#35.6
class Time:
    def __init__(self, hour, minute, second):
        self.hour = hour
        self.minute = minute
        self.second = second

    @staticmethod
    def is_time_valid(time_string):
        hour, minute, second = map(int, time_string.split(':'))
        return hour <= 24 and minute <= 59 and second <= 60
time_string = input()    

if Time.is_time_valid(time_string):
    t = Time.from_string(time_string)
    print(t.hour, t.minute, t.second)
else:
    print('잘못된 시간 형식입니다.')

    
```
