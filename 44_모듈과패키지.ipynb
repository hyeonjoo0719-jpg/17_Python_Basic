```python
# import로 모듈 가져오기
# import 모듈
# import 모듈1, 모듈2,
# 모듈.변수
# 모듈.함수()
# 모듈.클래스()omport math
import math
print(math.pi)
math.sqrt(4.0)
```

```python
## import as로 모듈 지정하기
# import 모듈 as 이름
import math as m
print(m.sqrt(4.0))
print(m.pi)
```

```python
# from import 로 모듈의 일부만 가져오기
# import math
from math import pi, sqrt
print(pi)
sqrt(4)
```

```python
# from 모듈 import *
from math import *
```

```python
# from import 로 모듈의 일부를 가져온 뒤에 이름 지정하기
# from 모듈 import 변수 as 이름
from math import sqrt as s
print(s(4.0))
print(s(2.0))
```
