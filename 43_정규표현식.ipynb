```python
# 문자열 판단하기
import re
re.match("Hello", "Hello, world")
```

```python
re.search("Hello", "Hello, World")
re.search("World!$", "Hello, World!")
```

```python
# 지정된 문자열이 하나라도 포함되어 있는지 판단하기
re.match("Hello|World", "Hello")
```

```python
# 범위 판단하기
print(re.match("[0-9]*", "1234"))
print(re.match("[0-9]+", "1234"))
print(re.match("[0-9]+", "abcd"))
```

```python
print(re.match("a*b", "b"))
print(re.match("a+b", "b"))
print(re.match("a+b", "aab"))
```

```python
# 문자가 한 개만 있는지 판단하기
print(re.match('abc?d', 'abd'))
print(re.match('ab[0-9]?c', 'ab3c'))
print(re.match('ab.d', 'abxd'))
```

```python
# 문자 개수 판단하기
# 문자{개수}
# (문자열){개수}

print(re.match('h{3}', "hhhello"))
print(re.match('(hello){3}', "hellohellohelloworld"))
```

```python
# [0-9]{개수}
print(re.match('[0-9]{3}-[0-9]{4}-[0-9]{4}', '010-1000-1000'))
print(re.match('[0-9]{3}-[0-9]{4}-[0-9]{4}', '010-1000-100'))
```

```python
# 숫자와 영문 문자를 조합해서 판단하기

print(re.match('[a-zA-Z0-9]+', 'Hello1234'))
print(re.match('[A-Z0-9]+', 'hello'))
```

```python
# 가-힣
re.match('[가-힣]+', '홍길동')
re.match('[가-힣]+ [가-힣]+', '대구시 달서구')
```

```python
## 특정 문자 범위에 포함되지 않는 지 판단하기
# [^범위]*
# [^범위]+

print(re.match("[^A-Z]+", "Hello"))
print(re.match("[^A-Z]+", "hello"))
```

```python
# ^[범위]*
# ^[범위]+

print(re.search('^[A-Z]+','Hello'))
```

```python
# [범위]*$
# [범위]+$

print(re.search('[0-9]+$', 'Hello1234'))
```

```python
#  특수 문자 판단하기
# \특수문자

print(re.search('\*+', '1 ** 2'))
print(re.match('[$()a-zA-Z0-9]+', '$(document)'))
```

```python
# \d -> [0-9]    # 모든 숫자
# \D -> [^0-9]   # 숫자를 제외한 모든 문자
# \w -> [a-zA-Z0-9_]  # 영문 대소문자, 숫자, 밑줄 문자
# \W -> [^a-zA-Z0-9_] # 영문 대소문자, 숫자, 밑줄 문자를 제외한 모든 문자

print(re.match('\d+', '1234'))
print(re.match('\D+', 'Hello'))
print(re.match('\w+', 'Hello_1234'))
print(re.match('\+', '(:)'))
```

```python
# 공백 처리하기
print(re.match('[a-zA-Z0-9 ]+', "Hello 1234"))
print(re.match(r'[a-zA-Z0-9 ]+', "Hello 1234"))
```

```python
# 그룹 사용하기
# (정규표현식) (정규표현식)
# 매치객체.group(그룹숫자)
m = re.match('([0-9]+) ([0-9]+)', '10 295')
print(m.group(1))   # 이거는 0이 아니고 1부터 시작함
print(m.group(2))
print(m.group(0))   # 전부 출력
print(m.groups())  # 각 그룹에 해당하는 문자열을 튜플 형태로 반환
```

```python
# 그룹 이름 정하기 -> (?P<이름>정규표현식)

m = re.match('(?P<func>[a-zA-Z_][a-zA-Z0-9_]+)\((?P<arg>\w+)\)', 'print(1234)')
print(m.group('func'))
print(m.group('arg'))
```

```python
# 패턴에 매칭되는 모든 문자열 가져오기
# re.findall("패턴", "문자열")
re.findall("[0-9]+", "1 2 Fizz 4 Buzz Fizz 7 8")
```

```python
# 문자열 바꾸기
# re.sub("패턴", "바꿀문자열", "문자열", 바꿀횟수)
print(re.sub("apple|orange", "fruit", "apple box orange tree"))
print(re.sub("[0-9]+", "n", "1 2 Fizz 4 Buzz Fizz 7 8"))
```

```python
# 교체함수('매치객체')
# re.sub("패턴", "교체함수", "문자열", 바꿀횟수)

def multiple10(m):
    n = int(m.group())
    return str(n * 10)

re.sub("[0-9]+", multiple10, "1 2 Fizz 4 Buzz Fizz 7 8")
```

```python
re.sub("[0-9]+", lambda m: str(int(m.group()) * 10), "1 2 Fizz 4 Buzz Fizz 7 8")
```

```python
#43.5

import re
p = re.compile('^[a-zA-Z0-9+-_.]+@[a-zA-Z0-9-]+\.[a-zA-Z0-9-.]+$')
emails = ['python@mail.example.com', 'python+kr@example.com',              # 올바른 형식
          'python-dojang@example.co.kr', 'python_10@example.info',         # 올바른 형식
          'python.dojang@e-xample.com',                                    # 올바른 형식
          '@example.com', 'python@example', 'python@example-com']          # 잘못된 형식

for email in emails:
    print(p.match(email) != None, end=' ')
```
