```python
# 31.4 회문 판별하기
def is_palindrome(word):
    if len(word) < 2:
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])

print(is_palindrome('level'))
```

```python
# 31.5
def fibonacci(n):
    a = 0
    b = 1
    for _ in range(n):         # 범위는 시작의 0부터가 0번째 피보나치 수가 되기에 n + 1이 아니고 n임.
        a, b = b , a + b
    return a

n = int(input())
print(fibonacci(n))
```
