```python
# 사람클래스로 학생 클래스 만들기
# class 기반클래스이름:
#     코드
# 
# class 파생클래스이름(기반클래스이름):
#     코드

class Person():
    def greeting(self):
        print('안녕하세요.')

class Student(Person):
    def study(self):
        print('공부하기')

james = Student()
james.greeting()
james.study()
```

```python
# 상속 관계와 포함 관계 알아보기
## 포함관계
class Person():
    def greeting(self):
        print('안녕하세요.')

class PersonList:
    def __init__(self):
        self.person_list = []

    def append_person(self, person):
        self.person_list.append(person)
        
```

```python
# 기반 클래스와 속성 사용하기
class Person:
    def __init__(self):
        print("Person__init__")
        self.hello = '안녕하세요'
        
class Student(Person):
    def __init__(self):
        print("Student__init__")
        self.school = "파이썬 코딩도장"

james = Student()
print(james.school)
print(james.hello)       # 뭐가 업데이트됨
```

```python
# super()로 기반 클래스 초기화하기

class Person:
    def __init__(self):
        print("Person__init__")
        self.hello = '안녕하세요'

class Student(Person):
    def __init__(self):
        print("Student__init__")
        super().__init__()  # 부모 클래스의 속성도 사용 가능해짐?
        self.school = "파이썬 코딩도장"    

print(james.school)
print(james.hello)    
```

```python
# 메서드 오버라이팅 사용하기
class Person:
    def greeting(self):
        print('안녕하세요')

class Student(Person):
    def greeting(self):
        print('안녕하세요. 저는 파이썬 코딩 도장 학생입니다.')

james = Student()
james.greeting()
james = Student()
```

```python


class Student(Person):
    def greeting(self):
        super().greeting # 부모 클래스의 안녕하세요르 가저ㅕ옴

        print('안녕하세요. 저는 파이썬 코딩 도장 학생입니다.')

james = Student()
james.greeting()
james = Student()
```
